```python
order_date = pd.to_datetime(df["Order Date"])
ship_date = pd.to_datetime(df["Ship Date"])

(ship_date < order_date).sum()
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])
```

```python
df[["Order Date", "Ship Date"]].dtypes
```

```python
df["Shipping Days"] = (
    df["Ship Date"] - df["Order Date"]
).dt.days

df["Shipping Days"].describe()
```

```python
df["Shipping Days"].value_counts().sort_index()
```

```python
df.groupby("Ship Mode")["Shipping Days"].agg(
    ["count", "mean", "min", "max"]
).sort_values("mean")
```

```python
df["Order ID"].duplicated().sum()
```

```python
print("Unique Orders:", df["Order ID"].nunique())
print("Unique Customers:", df["Customer ID"].nunique())
print("Unique Products:", df["Product ID"].nunique())
```

```python
order_sales = df.groupby("Order ID")["Sales"].sum()

order_sales.mean()
```

```python
print("Total Sales:", df["Sales"].sum())
print("Total Profit:", df["Profit"].sum())
print("Total Quantity:", df["Quantity"].sum())
```

```python
profit_margin = df["Profit"].sum() / df["Sales"].sum() * 100

print("Overall Profit Margin:", profit_margin, "%")
```

```python
df["Row ID"].nunique()
```

```python
print("Sales <= 0:", (df["Sales"] <= 0).sum())
print("Quantity <= 0:", (df["Quantity"] <= 0).sum())
print("Discount outside 0-1:", ((df["Discount"] < 0) | (df["Discount"] > 1)).sum())
```

```python
print("Order Date:", df["Order Date"].min(), "to", df["Order Date"].max())
print("Ship Date:", df["Ship Date"].min(), "to", df["Ship Date"].max())
```

```python
print("Loss-making rows:", (df["Profit"] < 0).sum())
print("Loss-making %:", (df["Profit"] < 0).mean() * 100)
```

```python
order_consistency = df.groupby("Order ID").agg(
    unique_customers=("Customer ID", "nunique"),
    unique_order_dates=("Order Date", "nunique")
)

print(
    "Orders with multiple customers:",
    (order_consistency["unique_customers"] > 1).sum()
)

print(
    "Orders with multiple order dates:",
    (order_consistency["unique_order_dates"] > 1).sum()
)
```
